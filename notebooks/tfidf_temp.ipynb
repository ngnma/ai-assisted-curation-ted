```python
# libraries
import sys
sys.path.append("..") 
from src.indexing.data import load_indexing_data
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.preprocessing import MultiLabelBinarizer
from sklearn.multiclass import OneVsRestClassifier
from sklearn.linear_model import LogisticRegression
import numpy as np

# data
data = load_indexing_data()
train_df = data.get('train')
val_df = data.get('val')

# train
mlb = MultiLabelBinarizer()
train_labels = mlb.fit_transform(train_df['labels']) # turn each talk's tag list into a 0/1 matrix with one column per tag.

tfidf = TfidfVectorizer()
train_features = tfidf.fit_transform(train_df['text']) # turn each talk's text into TF-IDF vectors.

multilabel_model = OneVsRestClassifier(LogisticRegression())
multilabel_model.fit(train_features, train_labels) # train a separate logistic regression model for each tag.

# validation
val_labels = mlb.transform(val_df['labels']) # turn each talk's tag list into a 0/1 matrix with one column per tag.
val_features = tfidf.transform(val_df['text']) # turn each talk's text into TF-IDF vectors.

val_proba = multilabel_model.predict_proba(val_features) # get probabilities for each class for each sample
idx = np.argsort(val_proba, axis=1, descending=True)  # Get indices of sorted probabilities in descending order
top_k = 10  # Define the number of top tags to retrieve
top_k_indices = idx[:, :top_k]  # Get the indices of the top-k tags for each row
top_k_tags = [[mlb.classes_[i] for i in row] for row in top_k_indices]  # Map indices to class names
```
